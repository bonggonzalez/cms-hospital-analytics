The analysis will investigate:

1. How does ED performance vary across U.S. hospitals?

## Relational Data Model Design

### Table Grains

**hospitals**
- Grain: One row per hospital/facility
- Primary key: `facility_id`
- `facility_id` uniqueness was validated against Hospital General Information.

**measures**
- Grain: One row per CMS measure
- Primary key: `measure_id`
- The dataset contains 30 unique Measure IDs, each corresponding to a unique Measure Name and Condition.

**hospital_measures**
- Grain: One measure observation per hospital per reporting period
- Candidate composite primary key:
  (`facility_id`, `measure_id`, `start_date`, `end_date`)
- This combination was tested against 138,084 source rows and produced zero duplicates.
- Foreign keys:
  - `facility_id` → `hospitals.facility_id`
  - `measure_id` → `measures.measure_id`

### Reporting Periods

The Timely and Effective Care dataset contains three distinct reporting periods. Therefore, reporting period must be part of the grain of `hospital_measures`.

### Unmatched Facilities

239 Facility IDs appear in Timely and Effective Care but not in Hospital General Information.

These records will not be discarded because the mismatch alone does not establish that they are invalid.

The ETL process will preserve these facilities using identifying information available in Timely and Effective Care. Hospital attributes unavailable from Hospital General Information will remain NULL.

A source/provenance field will distinguish hospitals originating from Hospital General Information from facilities found only in Timely and Effective Care.

### Modeling Principle

Hospital descriptive attributes belong in `hospitals`, measure definitions belong in `measures`, and hospital-specific measure results belong in `hospital_measures`.

This avoids unnecessary duplication while preserving the relationships required for analysis.

### Score Transformation

The CMS `Score` field contains both numeric and categorical values. Numeric measures require an analysis-ready numeric representation, while categorical measures such as `EDV` must retain their original values.

To preserve the source data while supporting numeric analysis:

- `Score` is retained as the original CMS-reported value.
- `score_numeric` is created using numeric conversion with `errors='coerce'`.
- Numeric scores are converted to numbers.
- Categorical values such as ED volume categories remain in `Score` and become missing (`NaN`) in `score_numeric`.
- `Not Available` remains preserved in `Score` and becomes `NaN` in `score_numeric`.

This approach preserves source traceability while providing a numeric field for quantitative analysis.

### Hospital Measures Table

The `hospital_measures` table stores hospital-level CMS measure observations.

- Grain: one measure observation per hospital per reporting period.
- Composite primary key: `facility_id`, `measure_id`, `start_date`, `end_date`.
- `facility_id` references the `hospitals` table.
- `measure_id` references the `measures` table.
- Raw score and sample values are retained for traceability.
- Numeric versions support quantitative analysis while non-numeric values become missing (`NaN`).
- Final table contains 138,084 observations.
- Composite-key validation found 0 duplicates.
- Foreign-key validation found 0 unmatched hospital IDs and 0 unmatched measure IDs.

```python
# Hospitals
hospitals['facility_id'] = hospitals['facility_id'].astype('string')
hospitals['zip_code'] = hospitals['zip_code'].astype('string')

# Measures
measures['measure_id'] = measures['measure_id'].astype('string')

# Hospital measures
hospital_measures['facility_id'] = hospital_measures['facility_id'].astype('string')
hospital_measures['measure_id'] = hospital_measures['measure_id'].astype('string')
```
